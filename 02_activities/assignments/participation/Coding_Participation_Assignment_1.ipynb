```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import scipy
import PIL
import requests

np.random.seed (613)
x=np.arange (50)
y=np.random.randint (0,100,50)
```

```python
fig,ax=plt.subplots (figsize=(5,3))
ax.scatter (x,y)

fig,ax= plt.subplots (figsize=(5, 3))
ax.plot (x,y)

fig,ax= plt.subplots (figsize=(5, 3))
ax.hist (y) #no need to input x values for histogram
ax.hist (y,edgecolor='black',bins=10) #customize histogram with edgecolor and bins
```

```python
font1={'family':'sans-serif','color':'blue','size':10}
font2={'family':'monospace','color':'darkred','size':8}

fig,ax= plt.subplots (figsize=(5, 3))
ax.plot (x,y)
ax.set_title ('Total growth over time', fontdict=font1)
ax.set_ylabel('Total growth', fontdict=font2)
ax.set_xlabel ('Years since start', fontdict=font2)
fig.tight_layout () #adjusts padding to prevent clipping of labels
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.scatter (x,
y,
marker='*',
color='green'
)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.grid(axis='y')
ax.grid(axis='y',color="grey",linewidth=1,linestyle='-.')

ax.plot (x,y,
marker='*',
color='#7425b9',
linestyle='--',
linewidth=2,
markeredgecolor = '#fa9359',
markerfacecolor = '#000000')
plt.show()
```
